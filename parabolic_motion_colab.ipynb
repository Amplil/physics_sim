```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
G = 9.8  # 地球上の重力加速度 [m/s²]
```

```python
# スライダー
v0_slider = widgets.FloatSlider(
    value=15.0,
    min=1.0,
    max=30.0,
    step=1.0,
    description="初速度:"
)

angle_slider = widgets.FloatSlider(
    value=45.0,
    min=0.0,
    max=90.0,
    step=1.0,
    description="角度:"
)

x_max_input = widgets.FloatText(
    value=100.0,
    description="X軸上限:"
)

y_max_input = widgets.FloatText(
    value=50.0,
    description="Y軸上限:"
)

output = widgets.Output()
```

```python
def draw(v0, angle, x_max, y_max):
    with output:
        clear_output(wait=True)

        # 角度をラジアンに変換
        theta = np.radians(angle)

        # 初速度を水平・鉛直方向に分解
        vx = v0 * np.cos(theta)
        vy = v0 * np.sin(theta)

        # 飛行時間
        if vy > 0:
            flight_time = 2 * vy / G
        else:
            flight_time = 0.1

        t = np.linspace(0, flight_time, 200)

        # 放物運動
        x = vx * t
        y = vy * t - 0.5 * G * t**2

        # 結果
        max_height = vy**2 / (2 * G)
        distance = x[-1]

        print(f"飛行時間   : {flight_time:.2f} s")
        print(f"最高点     : {max_height:.2f} m")
        print(f"水平到達距離: {distance:.2f} m")

        # グラフ
        plt.figure(figsize=(8, 5))
        plt.plot(x, y)
        plt.xlabel("x [m]")
        plt.ylabel("y [m]")
        plt.title("Projectile Motion")
        plt.grid()
        plt.xlim(0, x_max)
        plt.ylim(0, y_max)
        plt.show()


def update(change=None):
    draw(
        v0_slider.value,
        angle_slider.value,
        x_max_input.value,
        y_max_input.value,
    )
```

```python
# 値が変わったら再計算
v0_slider.observe(update, names="value")
angle_slider.observe(update, names="value")
x_max_input.observe(update, names="value")
y_max_input.observe(update, names="value")

# 表示
display(v0_slider)
display(angle_slider)
display(x_max_input)
display(y_max_input)
display(output)

# 最初のグラフ
update()
```
